```python
import csv
import os

def split_csv(file_path, rows_per_file=1000):
    
    base_dir = "D:/KHTN/HK6/Phân tích dữ liệu thông minh/"
    
    full_path = os.path.join(base_dir, file_path)


    file_base_name = os.path.splitext(file_path)[0]

    output_dir = os.path.join("data", "raw_data")
    os.makedirs(output_dir, exist_ok=True)

    with open(full_path, mode='r', encoding='utf-8', newline='') as infile:
        reader = csv.reader(infile)
        header = next(reader)

        rows = []
        file_count = 1
        for i, row in enumerate(reader, 1):
            rows.append(row)
            if i % rows_per_file == 0:
                
                output_file = os.path.join(output_dir, f"{file_base_name}_part{file_count}.csv")
                
                with open(output_file, mode='w', encoding='utf-8', newline='') as outfile:
                    writer = csv.writer(outfile)
                    writer.writerow(header)
                    writer.writerows(rows)
                print(f"✅ Đã tạo: {output_file} với {len(rows)} dòng")
                rows = []
                file_count += 1

        # ghi phần còn lại
        if rows:
            
            output_file = os.path.join(output_dir, f"{file_base_name}_part{file_count}.csv")
            with open(output_file, mode='w', encoding='utf-8', newline='') as outfile:
                writer = csv.writer(outfile)
                writer.writerow(header)
                writer.writerows(rows)
            print(f"✅ Đã tạo: {output_file} với {len(rows)} dòng")

# Ví dụ sử dụng (đường dẫn tương đối, file nằm cùng thư mục)
split_csv("movies.csv", rows_per_file=50000)
```
